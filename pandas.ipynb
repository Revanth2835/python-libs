# **loc**

syntax

**df.loc[si:ei:up,['col1',col2','col3'.....]]**

ending index is inclusive in loc




**loc is a Pandas label-based indexing method.**

**It is used to access rows and columns using their labels/names.**

syntax 2

**df.loc[row_label, column_label]**

```python
df
```

```python
print(df.loc[2]) #Select one row
print(df.loc[2, "quantity"]) #Select one column from a specific row
print(df.loc[:, ["quantity", "price"]]) #Select multiple columns
print(df.loc[[0, 2], ["quantity", "price"]]) #Select rows + columns
print(df.loc[df["quantity"] > 5]) #filtering with loc :========= df.loc[condition, columns]
print(df.loc[df["quantity"] > 5, ["products", "price"]]) #only particular columns
df.loc[4,"quantity"]=10  # modifying the data 
print(df.loc[4,"quantity"])
df.loc[df["quantity"] < 2, "quantity"] = 5 # modify multiple rows -- make quantity as 5 wher quantity less than 2
```

```python
df.loc[0:1,['products','quantity']]
```
