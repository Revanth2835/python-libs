# filtering on series

syntax

**var[condition]**

```python
sal=pd.Series([10000,35000,65000,55000,45000,40000],name='salary')
sal
```

```python
print(sal[sal<=50000])
print(sal[sal>10000])
print(sal[sal!=45000])
```

# 2 rules for multiple conditions 

**1. () for each and every conditions**

**2.  & for and ,| for or**



```python
sal
```

```python
print(sal[(sal>=55000) & (sal<=650000)])
```

```python
print(sal[(sal==55000) | (sal==65000)])
```

# **loc**

syntax

**df.loc[si:ei:up,['col1',col2','col3'.....]]**

ending index is inclusive in loc




**loc is a Pandas label-based indexing method.**

**It is used to access rows and columns using their labels/names.**

syntax 2

**df.loc[row_label, column_label]**

```python
df
```

```python
print(df.loc[2]) #Select one row
print(df.loc[2, "quantity"]) #Select one column from a specific row
print(df.loc[:, ["quantity", "price"]]) #Select multiple columns
print(df.loc[[0, 2], ["quantity", "price"]]) #Select rows + columns
print(df.loc[df["quantity"] > 5]) #filtering with loc :========= df.loc[condition, columns]
print(df.loc[df["quantity"] > 5, ["products", "price"]]) #only particular columns
df.loc[4,"quantity"]=10  # modifying the data 
print(df.loc[4,"quantity"])
df.loc[df["quantity"] < 2, "quantity"] = 5 # modify multiple rows -- make quantity as 5 wher quantity less than 2
```

```python
df.loc[0:1,['products','quantity']]
```

# **iloc**

syntax

**variable_name.iloc[row_slice,col_slice]**

**iloc means integer-location based indexing.**

**It is used to access DataFrame data using integer positions.**

```python
d={
    "movies_name":['toxic','mizapur','kgf','spiderman','kantara','ghilli'],
    'hero_names':['yash','kahlin bhaia','yash','tom holland','rishib shetty','vijay'],
    'heroine':['kiara','bhumi','srinihi','zendaya','sapthami','trisha'],
    "movies_name1":['toxic','mizapur','kgf','spiderman','kantara','ghilli'],
    'hero_names1':['yash','kahlin bhaia','yash','tom holland','rishib shetty','vijay'],
    'heroine1':['kiara','bhumi','srinihi','zendaya','sapthami','trisha'],
    "movies_name2":['toxic','mizapur','kgf','spiderman','kantara','ghilli'],
    'hero_names2':['yash','kahlin bhaia','yash','tom holland','rishib shetty','vijay'],
    'heroine2':['kiara','bhumi','srinihi','zendaya','sapthami','trisha']
}
movies=pd.DataFrame(d)
movies
```

```python
movies.iloc[0:4,0:6]
```

```python
movies.iloc[::2,::2]
```

```python
print("=======select one row======")
print(movies.iloc[2]) 

print("=========one cell=========")
print(movies.iloc[2,0]) 

print("=========select multiple rows=========")
print(movies.iloc[[0,2]])

print("=============Select multiple columns=========")
print("============select all rows , only columns at 1 and 3=============")
print(movies.iloc[:,[1,3]])

print("=================select all rows , select columns from 1 to 3(columns 1 and 2)")
print(movies.iloc[:,1:3])
```

```python
print(movies.iloc[:,[1,3]])

print(movies.iloc[:,1:3])
```
